```python
import random
def dataset_len():
  return len(english)


def get_random_word():
  len = dataset_len()
  r = random.randint(0,len-1)
  return english[r],chinese[r]


def get_word(idx):
  return english[idx],chinese[idx]


def gen_que(idx):
  #生成除了idx外的三個選項，並和正確選項包在一起回傳
  ans = get_word(idx)
  pool = set(range(0,dataset_len()))
  pool.remove(idx)
  options = random.sample(sorted(pool),3)
  options.append(idx)
  random.shuffle(options)
  return idx,options
```

```python
print("單字總數：", dataset_len())
en, zh = get_random_word()
print("隨機單字：", en)
print("中文解釋：", zh)
```

```python
order = list(range(0,dataset_len()))
random.shuffle(order)

correct = 0
wrong = 0

for i in order:
  e,c = get_word(i)

  print(f"{e}的中文是以下哪個?")
  ans,option = gen_que(i)
  print(chinese[option[0]],chinese[option[1]],chinese[option[2]],chinese[option[3]],sep=',')
  player_choice = int(input())
  # print(option[player_choice],i)
  if option[player_choice] == i:
    print("正確")
    correct+=1
  else:

    print("錯誤")
    wrong+=1

print(f"正確:{correct}, 錯誤:{wrong}")
```
